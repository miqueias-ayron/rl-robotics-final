# Deterministic Taxi Driver: Sarsa Algorithm

Treinamento de um agente Sarsa no Taxi-v4 determinístico.

Serão realizadas cinco execuções independentes, utilizando os mesmos
hiperparâmetros do Q-learning

```python
import sys
from pathlib import Path

import gymnasium as gym
import numpy as np

# Localiza a raiz do projeto a partir da pasta de execução do notebook.
project_root = next(
    path
    for path in (Path.cwd(), *Path.cwd().parents)
    if (path / "algorithms").is_dir()
)

sys.path.insert(0, str(project_root))

from algorithms.Sarsa import Sarsa

data_dir = project_root / "data"
data_dir.mkdir(exist_ok=True)
```

```python
alpha = 0.5
gamma = 0.7
epsilon = 0.2
epsilon_min = 0.05
epsilon_dec = 0.99
episodes = 10000
executions = 5
```

```python
all_rewards = []
q_tables = []

for execution in range(1, executions + 1):
    print(f"\nTreinamento Sarsa: {execution}/{executions}")

    env = gym.make("Taxi-v4")

    agent = Sarsa(
        env,
        alpha=alpha,
        gamma=gamma,
        epsilon=epsilon,
        epsilon_min=epsilon_min,
        epsilon_dec=epsilon_dec,
        episodes=episodes,
    )

    filename = data_dir / "q-table-sarsa-deterministic.csv"

    q_table, rewards = agent.train(str(filename))

    q_tables.append(q_table)
    all_rewards.append(rewards)

    env.close()

all_rewards = np.array(all_rewards)
q_tables = np.array(q_tables)
```
